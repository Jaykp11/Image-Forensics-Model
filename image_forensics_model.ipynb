```python
import os
import cv2
import random
import numpy as np

REAL_DIR = "dataset/real"
FAKE_DIR = "dataset/fake"

os.makedirs(FAKE_DIR, exist_ok=True)

def add_noise(image):
    noise = np.random.normal(0, 25, image.shape).astype(np.int16)
    noisy = image.astype(np.int16) + noise
    return np.clip(noisy, 0, 255).astype(np.uint8)

def blur_image(image):
    return cv2.GaussianBlur(image, (5, 5), 0)

def adjust_brightness(image):
    value = random.randint(-50, 50)
    hsv = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)

    v = hsv[:, :, 2].astype(np.int16)
    v = np.clip(v + value, 0, 255)

    hsv[:, :, 2] = v.astype(np.uint8)
    return cv2.cvtColor(hsv, cv2.COLOR_HSV2BGR)

def adjust_contrast(image):
    alpha = random.uniform(0.5, 1.5)
    return cv2.convertScaleAbs(image, alpha=alpha, beta=0)

print("Creating fake dataset...")

for category in os.listdir(REAL_DIR):
    real_path = os.path.join(REAL_DIR, category)

    if not os.path.isdir(real_path):
        continue

    fake_path = os.path.join(FAKE_DIR, category)
    os.makedirs(fake_path, exist_ok=True)

    for img_name in os.listdir(real_path):
        img_path = os.path.join(real_path, img_name)

        image = cv2.imread(img_path)
        if image is None:
            continue

        # Random tampering selection
        choice = random.choice(["noise", "blur", "brightness", "contrast"])

        if choice == "noise":
            tampered = add_noise(image)
        elif choice == "blur":
            tampered = blur_image(image)
        elif choice == "brightness":
            tampered = adjust_brightness(image)
        else:
            tampered = adjust_contrast(image)

        save_path = os.path.join(fake_path, img_name)
        cv2.imwrite(save_path, tampered)

print("✅ Fake dataset created successfully!")
```

```python
import os
import pandas as pd

REAL_DIR = "dataset/real"
FAKE_DIR = "dataset/fake"

data = []

# Real images (label = 1)
for category in os.listdir(REAL_DIR):
    category_path = os.path.join(REAL_DIR, category)

    for img in os.listdir(category_path):
        img_path = os.path.join(category_path, img)
        data.append([img_path, 1])

# Fake images (label = 0)
for category in os.listdir(FAKE_DIR):
    category_path = os.path.join(FAKE_DIR, category)

    for img in os.listdir(category_path):
        img_path = os.path.join(category_path, img)
        data.append([img_path, 0])

# Create dataframe
df = pd.DataFrame(data, columns=["image_path", "label"])

# Shuffle dataset
df = df.sample(frac=1).reset_index(drop=True)

# Save CSV
df.to_csv("dataset.csv", index=False)

print("✅ Dataset CSV created!")
print(df.head())
```

```python
from sklearn.model_selection import train_test_split

train_df, test_df = train_test_split(df, test_size=0.2, random_state=42)

print("Train size:", len(train_df))
print("Test size:", len(test_df))
```

```python
from torch.utils.data import Dataset
import torch
from PIL import Image

class ImageDataset(Dataset):
    def __init__(self, dataframe, transform=None):
        self.df = dataframe
        self.transform = transform

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        img_path = self.df.iloc[idx]["image_path"]
        label = self.df.iloc[idx]["label"]

        image = Image.open(img_path).convert("RGB")

        if self.transform:
            image = self.transform(image)

        return image, torch.tensor(label)
```
